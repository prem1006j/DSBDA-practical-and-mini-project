```python
# Assignment 3: Descriptive Statistics

import numpy as np
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

# ── Load Iris Dataset ─────────────────────────────────────────────
df = pd.read_csv('dataset/iris.csv')
# Column names: sepal_length, sepal_width, petal_length, petal_width, species
print("Iris dataset loaded.")
print(df.head())
print(f"\nShape: {df.shape}")
print(f"\nSpecies: {df['species'].unique()}")

# ── 1. Summary Statistics Grouped by Species ──────────────────────
print("\n── Summary Statistics Grouped by Species ──")
grouped = df.groupby('species')[['sepal_length','sepal_width',
                                  'petal_length','petal_width']].agg(
    ['mean','median','min','max','std']
)
print(grouped.to_string())

# Grouped mean
print("\n── Group Means ──")
print(df.groupby('species').mean().round(3))

# ── 2. Percentile, Mean, Std for Each Species ─────────────────────
species_list = ['setosa', 'versicolor', 'virginica']
features     = ['sepal_length', 'sepal_width', 'petal_length', 'petal_width']

for sp in species_list:
    subset = df[df['species'] == sp]
    print(f"\n{'='*50}")
    print(f"Species: Iris-{sp} (n={len(subset)})")
    print(f"{'='*50}")
    for feat in features:
        vals = subset[feat]
        print(f"\n  {feat}:")
        print(f"    Mean:   {vals.mean():.4f}")
        print(f"    Median: {vals.median():.4f}")
        print(f"    Std Dev:{vals.std():.4f}")
        print(f"    Min:    {vals.min():.4f}")
        print(f"    Max:    {vals.max():.4f}")
        print(f"    25th %: {vals.quantile(0.25):.4f}")
        print(f"    50th %: {vals.quantile(0.50):.4f}")
        print(f"    75th %: {vals.quantile(0.75):.4f}")

# ── 3. Full describe() per species ───────────────────────────────
print("\n── describe() per species ──")
for sp in species_list:
    print(f"\nIris-{sp}:")
    print(df[df['species']==sp][features].describe().round(3))

# ── 4. Visualization ─────────────────────────────────────────────
fig, axes = plt.subplots(2, 2, figsize=(12, 8))
for i, feat in enumerate(features):
    ax = axes[i//2][i%2]
    for sp in species_list:
        vals = df[df['species']==sp][feat]
        ax.hist(vals, bins=15, alpha=0.6, label=f'Iris-{sp}')
    ax.set_title(f'Distribution: {feat}')
    ax.set_xlabel(feat)
    ax.set_ylabel('Frequency')
    ax.legend(fontsize=8)

plt.suptitle('Iris Dataset — Feature Distributions by Species', fontsize=13)
plt.tight_layout()
plt.savefig('screenshots/distributions.png', dpi=120, bbox_inches='tight')
plt.show()

print("\nDescriptive Statistics Assignment Complete ✓")
```
