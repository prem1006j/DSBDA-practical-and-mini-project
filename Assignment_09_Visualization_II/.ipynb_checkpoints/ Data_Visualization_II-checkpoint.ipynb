```python
import seaborn as sns
import matplotlib.pyplot as plt

df = sns.load_dataset('titanic')
```

```python
sns.boxplot(x='sex', y='age', hue='survived', data=df)
plt.title("Age Distribution by Gender and Survival")
plt.xlabel("Gender")
plt.ylabel("Age")
plt.show()
```

```python
sns.boxplot(x='sex', y='age', hue='survived', data=df, palette='Set2')
plt.title("Age Distribution by Gender & Survival")
plt.show()
```

```python
sns.violinplot(x='sex', y='age', hue='survived', data=df, split=True)
plt.title("Violin Plot of Age Distribution")
plt.show()
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

df = sns.load_dataset('titanic')

plt.figure(figsize=(8,6))
sns.boxplot(x='sex', y='age', hue='survived', data=df, palette='Set3')
plt.title("Age Distribution by Gender and Survival")
plt.xlabel("Gender")
plt.ylabel("Age")
plt.legend(title="Survived")
plt.show()
```

```python
plt.figure(figsize=(8,6))
sns.boxplot(x='sex', y='age', hue='survived', data=df)
sns.swarmplot(x='sex', y='age', hue='survived', data=df, dogde=True, color='black')
plt.title("Age Distribution with Individual Points")
plt.show()
```

```python
sns.catplot(x='sex', y='age', hue='survived', 
            kind='box', data=df, height=5, aspect=1.5)
plt.title("Age vs Gender vs Survival")
plt.show()
```

```python
print(df.groupby(['sex', 'survived'])['age'].mean())
```
